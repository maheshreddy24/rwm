```python
import os
import csv
import random
from decord import VideoReader, cpu

# Dataset path
video_dir = "/home/rvm/datasets/ssv2/20bn-something-something-v2"

# Output CSVs
train_csv = "/home/rvm/datasets/ssv2/train.csv"
test_csv = "/home/rvm/datasets/ssv2/test.csv"

# Collect all video files
video_paths = [
    os.path.join(video_dir, f)
    for f in os.listdir(video_dir)
    if f.lower().endswith((".webm", ".mp4", ".avi", ".mkv"))
]

print(f"Found {len(video_paths)} videos.")

# Shuffle and split
random.seed(42)
random.shuffle(video_paths)

split_idx = int(0.9 * len(video_paths))
train_videos = video_paths[:split_idx]
test_videos = video_paths[split_idx:]


def write_csv(video_list, csv_path):
    with open(csv_path, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["path", "num_frames", "fps", "duration_sec"])

        for i, video_path in enumerate(video_list):
            try:
                vr = VideoReader(video_path, ctx=cpu(0), num_threads=1)

                num_frames = len(vr)
                fps = vr.get_avg_fps()
                duration = num_frames / fps if fps > 0 else 0

                writer.writerow([
                    video_path,
                    num_frames,
                    round(fps, 2),
                    round(duration, 2)
                ])

                if (i + 1) % 100 == 0:
                    print(f"{os.path.basename(csv_path)}: {i + 1}/{len(video_list)} processed")

            except Exception as e:
                print(f"Failed: {video_path}")
                print(e)

    print(f"Saved to {csv_path}")


write_csv(train_videos, train_csv)
write_csv(test_videos, test_csv)

print(f"Train: {len(train_videos)} videos")
print(f"Test: {len(test_videos)} videos")


```
