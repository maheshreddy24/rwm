```python
import csv
import json
import os

video_dir = "/home/rvm/datasets/ssv2/20bn-something-something-v2/"

# ---------- Create SSV2 validation CSV ----------
with open("ssv2/labels/validation.json", "r") as f:
    val_data = json.load(f)

test_csv = "test_dataset_ssv2_kine.csv"

with open(test_csv, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["path", "source"])

    for item in val_data:
        writer.writerow([
            os.path.join(video_dir, f"{item['id']}.webm"),
            "ssv2"
        ])

# ---------- Create SSV2 train CSV ----------
with open("ssv2/labels/train.json", "r") as f:
    train_data = json.load(f)

train_csv = "train_dataset_ssv2_kine.csv"

with open(train_csv, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["path", "source"])

    # SSV2 samples
    for item in train_data:
        writer.writerow([
            os.path.join(video_dir, f"{item['id']}.webm"),
            "ssv2"
        ])

    # Kinetics samples
    with open("kinetics_train_set_clip_paths.csv", "r", newline="") as kf:
        reader = csv.DictReader(kf)

        for row in reader:
            writer.writerow([row["clip_path"], "kinetics"])
```

```python
import os
import csv
import random

dir = '/home/ego4d/data/v2/video_540ss/'

video_paths = [
    os.path.join(dir, p)
    for p in os.listdir(dir)
    if p.endswith(".mp4")
]

# Shuffle for random train/test split
random.seed(42)  # reproducible split
random.shuffle(video_paths)

# 10% test, 90% train
test_size = int(0.10 * len(video_paths))

test_paths = video_paths[:test_size]
train_paths = video_paths[test_size:]

# Write train CSV
with open('ego4d_train.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['path'])
    for path in train_paths:
        writer.writerow([path])

# Write test CSV
with open('ego4d_test.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['path'])
    for path in test_paths:
        writer.writerow([path])

print(f"Total videos: {len(video_paths)}")
print(f"Train videos: {len(train_paths)} ({len(train_paths)/len(video_paths)*100:.1f}%)")
print(f"Test videos:  {len(test_paths)} ({len(test_paths)/len(video_paths)*100:.1f}%)")
```
